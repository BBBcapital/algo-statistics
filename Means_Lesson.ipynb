## 1. The arithmetic mean

The arithmetic mean is the sum of a collection of numbers divided by the count of those numbers, representing the standard average of a dataset.

The arithmetic mean is used very frequently to summarize numerical data, and is usually what is assumed to be meant by the term "average." It is defined as the sum of the observations divided by the number of observations:
$$\mu = \frac{\sum_{i=1}^N X_i}{N}$$

where $X_1, X_2, \ldots , X_N$ are our observations.

In the next cell we will go through an example.

```python
# Two useful statistical libraries
from collections import Counter
import scipy.stats as stats
import numpy as np

values = [1, 2, 4, 4, 5, 6, 6, 8]
v2 = values + [100]

total = sum(values)
number_of_values = len(values)
arithmetic_mean = total / number_of_values

print("Total:", total)
print("Number of values:", number_of_values)
print("Arithmetic mean:", arithmetic_mean)

print("Mean of v2:", sum(v2), '/', len(v2), '=', np.mean(v2))
```

## 2. A weighted mean

A weighted mean lets some values count more than others.

A <i>weighted</i> arithmetic mean, is useful for explicitly specifying the number of times each observation should be counted. For instance, in computing the average value of a portfolio, it is more convenient to say that 70% of your stocks are of type X rather than making a list of every share you hold. Hence, in a portfolio, the weights are the shares of money invested in each asset.

The weighted arithmetic mean is defined as
$$\sum_{i=1}^n w_i X_i $$

where $\sum_{i=1}^n w_i = 1$. In the usual arithmetic mean, we have $w_i = 1/n$ for all $i$.

Suppose 75 percent of the portfolio earns 6 percent and 25 percent loses 2 percent. The portfolio return is the sum of each return times its weight.


```python
asset_returns = [6, -2]
portfolio_weights = [0.75, 0.25]

if len(portfolio_weights) != len(asset_returns):
    raise ValueError("The number of weights must match the number of returns.")

weighted_mean = sum(
    weight * asset_return
    for weight, asset_return in zip(portfolio_weights, asset_returns)
)

print("Weights add to:", sum(portfolio_weights))
print("Weighted portfolio return:", weighted_mean, "%")
```

The result is 4 percent. This calculation assumes the weights describe the portfolio at the start of the period.

## 3. The median and the mode

The median is the middle value after sorting the data in increasing order. When we have an odd number $n$ of data points, this is simply the value in position $(n+1)/2$. When we have an even number of data points, the list splits in half and there is no item in the middle; so we define the median as the average of the two middle values.

The median is less affected by extreme values in the data than the arithmetic mean. It tells us the value that splits the data set in half, but not how much smaller or larger the other values are.

The mode is the value that appears most often. A list can have more than one mode, or no repeated value. The mode can be applied to non-numerical data, unlike the mean and the median. One situation in which it is useful is for data whose possible values are independent. For return data with many unique values, we can group returns into ranges and look for the most common range.

Median and mode are other ways to describe a centre. They are not types of mean.


```python
values = [1, 2, 4, 4, 5, 6, 6, 8]
v2 = values + [100]

counts = Counter(values)
most_common_count = max(counts.values())
if most_common_count > 1:
    modes = [
        value
        for value, count in counts.items()
        if count == most_common_count
    ]
else:
    modes = []

print("Sorted values:", sorted(values))
print("Sorted v2:", sorted(v2))
print("Median of values:", np.median(values))
print("Median of v2:", np.median(v2))
print("Counts by value:", counts)
if modes:
    print("Mode(s):", modes, "appear", most_common_count, "times each")
else:
    print("Mode: no repeated value")
```

## 4. The geometric mean

While the arithmetic mean averages using addition, the geometric mean uses multiplication:
$$ G = \sqrt[n]{X_1X_2\cdots X_n} $$

for observations $X_i \geq 0$. We can also rewrite it as an arithmetic mean using logarithms:
$$ \ln G = \frac{\sum_{i=1}^n \ln X_i}{n} $$

The geometric mean is always less than or equal to the arithmetic mean (when working with non-negative observations), with equality only when all of the observations are the same.

Returns compound because each return acts on the value left after the previous return. A 10 percent gain followed by a 10 percent loss does not return the investment to its starting value.

Starting with £100:
- A 10 percent gain gives £110.
- A 10 percent loss on £110 leaves £99.

The geometric mean gives the constant return per period that would produce the same overall growth. For returns, multiply the growth factors, take the nth root, then subtract 1. Each return must be greater than -100 percent.


```python
import math

period_returns = [0.10, -0.10]
growth_factors = [1 + rate for rate in period_returns]

ending_value = 100 * math.prod(growth_factors)
geometric_mean_return = math.prod(growth_factors) ** (1 / len(growth_factors)) - 1
arithmetic_mean_return = sum(period_returns) / len(period_returns)

print("Ending value from £100:", round(ending_value, 2))
print("Arithmetic mean return:", round(arithmetic_mean_return * 100, 2), "%")
print("Geometric mean return per period:", round(geometric_mean_return * 100, 2), "%")

# Use scipy's gmean function to compute the geometric mean
print("Geometric mean of values:", stats.gmean(values))
print("Geometric mean of v2:", stats.gmean(v2))
```

The arithmetic mean is 0 percent, but the value falls from £100 to £99. The geometric mean captures that compound growth rate, which is about -0.50 percent per period.

## 5. The harmonic mean

The harmonic mean can be used when the data can be naturally phrased in terms of ratios. For instance, in the dollar-cost averaging strategy, a fixed amount is spent on shares of a stock at regular intervals. The higher the price of the stock, then, the fewer shares an investor following this strategy buys. The average (arithmetic mean) amount they pay for the stock is the harmonic mean of the prices.

For example if you spend £60 on shares at £10 each, then spend another £60 when the share price is £20. The first purchase buys 6 shares and the second buys 3. You spent £120 and bought 9 shares, so your average cost per share is £13.33.


```python
prices = [10, 20]
cash_spent_each_time = 60

shares_bought = [cash_spent_each_time / price for price in prices]
total_cash = cash_spent_each_time * len(prices)
total_shares = sum(shares_bought)
average_cost_per_share = total_cash / total_shares

harmonic_mean_price = len(prices) / sum(1 / price for price in prices)

print("Shares bought at each price:", shares_bought)
print("Average cost per share:", round(average_cost_per_share, 2))
print("Harmonic mean of prices:", round(harmonic_mean_price, 2))

print("Harmonic mean of values:", stats.hmean(values))
print("Harmonic mean of v2:", stats.hmean(v2))
```

The harmonic mean gives the average price per share in this equal-cash example. The arithmetic mean of £10 and £20 is £15, which does not match the actual average cost per share here.

## Main idea

Choose the summary that fits the question:

- Arithmetic mean: each observation counts equally.
- Weighted mean: observations have different shares or importance.
- Median: the middle value after sorting.
- Mode: the most common value or group.
- Geometric mean: compound growth across time.
- Harmonic mean: some ratios and equal-cash price examples.

A single summary hides detail. When you study financial returns, also look at the observations, the time period and a measure of spread.
